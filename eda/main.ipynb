### Aufgabe 3: Was bedeuten diese Variablen? In welchen Einheiten wurden sie gemessen?

```python
for i in df_penguins.columns:
    if len(i.split("_")) == 3:
        display(i.split("_")[0] + ":" + i.split("_")[1] + "[" + i.split("_")[2] + "]")
    elif len(i.split("_")) == 2:
        display(i.split("_")[0] + ":" + i.split("_")[1])
    else:
        display(i.split("_")[0])
```
